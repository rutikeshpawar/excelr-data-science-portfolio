 # create a system in which student will come and enter marks from 1-100
 # you need to classify the user into correct grade based on the below grading system
 # between 80-100 Grade A
 # between 60-79 Grade B
 # between 35-59 Grade C
 # below 35 is Fail
 ## NO NEGATIVE MARKS OR MARKS GREATER THAN 100 CAN BE ENTERED 
 ## when user try to do so throw a warning message

```python
try:
    marks = int(input("Enter your marks"))
    if marks >= 80 and marks <= 100 :
        print("Grade A")
    elif marks >= 60 and marks <= 79:
        print("Grade B")
    elif marks >= 35 and marks <= 59:
        print("Grade C")
    elif marks < 35 and marks >= 0:
        print("Fail")
    else: 
        print("Please enter valid marks")
except:
    print(f"{marks} is not valid Input")
```

```python
# Function that will take list as a input and return the average of that list
def getAverage(mylist:list):
    output = sum(mylist)/len(mylist)
    return output

getAverage([1,2,3,4,5,6,7,8,9])
```

```python
# GetSquareAvg
def getSqAvg(mylist:list):
    var = getAverage(mylist)
    output = var**2
    return output

getSqAvg([1,2,3,4,5,6,7,8,9])
```

```python
# both get squared avegara and average of list
def mylogic(mylist:list):
    avg = getAverage(mylist)
    sq_avg = getSqAvg(mylist)

    return {
        "average" : avg,
        "Square_average" : sq_avg
    }
```
