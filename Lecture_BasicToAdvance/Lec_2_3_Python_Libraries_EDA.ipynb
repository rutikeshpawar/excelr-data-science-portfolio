```python
# Null values in the dataset
data_sal.isnull().sum()
```

```python
# mean  -Used on Numeric Data, - Avegage value eg. Range 1 - 100 so mean is 50
# median - Used on Numeric Data, - Medium value eg. Range 1 - 100 so median is 50
# mode - Used on Categorical Data, - Most repeated value eg. Range 1 - 100 so mode is 1
mean_salary = data_sal["Salary"].mean()
round(mean_salary, 2)
```

```python
data_sal["Salary"].fillna(mean_salary, inplace=True)
```

```python
mode_companyname = data_sal["Company Name"].mode()[0]
mode_companyname
```

```python
data_sal["Company Name"].fillna(mode_companyname[0], inplace=True)
```

```python
mode_jobtitle = data_sal["Job Title"].mode()[0]
mode_jobtitle
```

```python
data_sal["Job Title"].fillna(mode_jobtitle[0], inplace=True)
```

```python
mode_location = data_sal["Location"].mode()[0]
mode_location
```

```python
data_sal["Location"].fillna(mode_location, inplace=True)
```

```python
mode_empstatus = data_sal["Employment Status"].mode()[0]
mode_empstatus
```

```python
data_sal["Employment Status"].fillna(mode_empstatus, inplace=True)
```

```python
mode_jobroles = data_sal["Job Roles"].mode()[0] 
mode_jobroles
```

```python
data_sal["Job Roles"].fillna(mode_jobroles, inplace=True)
```
