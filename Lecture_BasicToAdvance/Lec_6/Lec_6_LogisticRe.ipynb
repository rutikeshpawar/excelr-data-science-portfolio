```python
import pandas as pd

def fillNullValues(data):
    try:
        # Loop through all columns
        for i in data.columns:
            # 1. Corrected: Decimal check for 20% missing values
            if data[i].isnull().mean() > 0.20:
                data.drop(columns=[i], inplace=True) # Corrected drop syntax
                continue # Skip to next column since this one is deleted
                
            # 2. Handle Text / Categorical data
            elif data[i].dtype == 'object':
                column_mode = data[i].mode()
                if not column_mode.empty:
                    data[i].fillna(column_mode[0], inplace=True)
                else:
                    data[i].fillna("Unknown", inplace=True)
                    
            # 3. Handle Numerical data
            else:
                theRange = data[i].max() - data[i].min()
                if theRange > 200:
                    data[i].fillna(data[i].mean(), inplace=True)
                else:
                    data[i].fillna(data[i].median(), inplace=True)
                    
        # Corrected: Return statement moved OUTSIDE the for loop
        return {
            "message" : "Null values filled successfully"
        }
        
    except Exception as e:
        return {
            "message" : "Error occurred while filling null values",
            "error" : str(e)
        }

# Run the function on your dataset
result = fillNullValues(data_new)
print(result)
```

```python
X = data_new.values[:, 0:-1]
Y = data_new.values[:, -1]
```

```python
print(X.shape)
print(Y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=10)
```

```python
from sklearn.linear_model import LogisticRegression
# Create a Model Object
classifier = LogisticRegression()

# Train the model Objest
classifier.fit(X_train, Y_train)
```

```python
Y_pred = classifier.predict(X_test)
print("Predicted Values:\n", Y_pred)
print(list(zip(Y_test, Y_pred)))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

cfm = confusion_matrix(Y_test, Y_pred)
print("Confusion Matrix:\n", cfm, "\n") 

print("Classification Report:")
print(classification_report(Y_test, Y_pred))

acc = accuracy_score(Y_test, Y_pred)
print("Accuracy Score:", acc)
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix, classification_report

# 1. Build a Pipeline to scale features and handle class imbalance
pipeline = Pipeline([
    ('scaler', StandardScaler()),  # Fixes the convergence warnings
    ('lr', LogisticRegression(class_weight='balanced', solver='lbfgs', max_iter=5000))
])

# 2. Set up the parameter grid tuning (note the double underscore 'lr__C')
param_grid = {'lr__C': [0.01, 0.1, 1, 10, 100]}

# 3. Optimize using GridSearchCV targeting F1-Score
grid_search = GridSearchCV(estimator=pipeline, param_grid=param_grid, scoring='f1', cv=5)
grid_search.fit(X_train, Y_train)

# 4. Extract the optimized model
best_model = grid_search.best_estimator_
print(f"Best Parameters Found: {grid_search.best_params_}\n")

# 5. Predict probabilities and apply a custom threshold (0.4) to catch more class 1s
Y_prob = best_model.predict_proba(X_test)[:, 1]
custom_threshold = 0.48
Y_pred_custom = (Y_prob >= custom_threshold).astype(int)

# 6. Print your updated results
print("--- Optimized Performance Metrics ---")
print("Adjusted Confusion Matrix:\n", confusion_matrix(Y_test, Y_pred_custom))
print("\nAdjusted Classification Report:\n", classification_report(Y_test, Y_pred_custom))
```
